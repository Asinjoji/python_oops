Create a class circle with a constructor initializing the radius r, another func to cal the area of the circle, another fun for circumference of the circle.

```python
class circle:
  def __init__(self,r):
    self.radius=r
  def area(self):
    return 3.14*self.radius*self.radius
  def circumference(self):
    return 2*3.14*self.radius
  def displaydata(self):
    print(f"Area : {self.area()} and Circumference : {self.circumference()}")
c1=circle(5)
c1.displaydata()
```

There ar 2 classes student and masterstudent. Student is the base class and masterstudent is the derived class. The class student has 2 member variables ID and Mark. There is a constructor to intitialize these variables and a function show_result() which will check the mark for passed and failed. If the mark is >= 40 it is a pass otherwise it is a fail. Similarly in the class masterstudent contain a function show_result() that will check whether the mark is >= 50. If that  is the case then it is a pass, otherwise it will be a fail. Initialize a student object and a masterstudent object and call the functions respectively

```python
class Student:
    def __init__(self, id, mark):
        self.id = id
        self.mark = mark
    def show_result(self):
        if self.mark >= 40:
            print("Student ID:", self.id, "- Pass")
        else:
            print("Student ID:", self.id, "- Fail")
class Master_Student(Student):
    def show_result(self):
        if self.mark >= 50:
            print("Master Student ID:", self.id, "- Pass")
        else:
            print("Master Student ID:", self.id, "- Fail")
student_obj = Student(101, 45)
master_student_obj = Master_Student(102, 55)
student_obj.show_result()
master_student_obj.show_result()
```

```python
class Polygon:
  def __init__(self):
    pass
  def area(self):
    pass
  def perimeter(self):
    pass
class Rectangle(Polygon):
  def __init__(self,l,b):
    self.length=l
    self.breadth=b
  def area(self):
    return self.length*self.breadth
  def perimeter(self):
    return(2*(self.length+self.breadth))
class Triangle(Polygon):
  def __init__(self,a,b,c):
    self.p=a
    self.q=b
    self.r=c
    s=(self.p+self.q+self.r)/2
    return((s*(s-self.p)*(s-self.q)*(s-self.r))**0.5)
  def perimeter(self):
    return(self.p+self.q+self.r)
r1=Rectangle(5,10)
print(r1.area())
print(r1.perimeter())
t1=Triangle(3,3,3)
print(t1.area())
print(t1.perimeter())
```
